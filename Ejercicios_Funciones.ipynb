##Problema 1: Juego de "Piedra, Papel o Tijera"



Diseña un programa que permita al usuario jugar "Piedra, Papel o Tijera" contra la computadora. El
programa debe seguir estos pasos:
1. Al iniciar, se debe generar un número aleatorio entre 1 y 3:
o 1 significa que la computadora eligió "Piedra".
o 2 significa que la computadora eligió "Papel".
o 3 significa que la computadora eligió "Tijera". (No se muestra la elección de la
computadora todavía).
2. El usuario debe ingresar su elección: "Piedra", "Papel" o "Tijera".
3. Luego, se muestra la elección de la computadora.
4. El programa debe indicar quién es el ganador, siguiendo estas reglas:
o Piedra vence a Tijera (la rompe).
o Tijera vence a Papel (lo corta).
o Papel vence a Piedra (lo envuelve).
o Si ambos jugadores eligen lo mismo, el juego debe repetirse hasta que haya un
ganador.
5. Agrega un sistema de puntaje. El programa debe llevar un registro de cuántas veces gana el
usuario y cuántas veces gana la computadora.
6. Al finalizar cada ronda, se muestra el puntaje actual y se pregunta al usuario si desea seguir
jugando. Si el usuario responde "sí", el juego continúa. Si responde "no", el programa muestra

````python
import random

#Mensaje de entrada
print("Bienvenido al juego Piedra, Papel o Tijera.")
print("Tendrás que elegir entre: 'piedra', 'papel' o 'tijera'. ¡Buena suerte!\n")

#Inicialización de las variables de los puntos
puntos_usuario = 0
puntos_cpu = 0

play_again = True

#Funciones a implementar
#Función para que la CPU elija
def juegos():
    return random.randint(1, 3)

#Función para la elección del usuario
def game():
    choice_user = input("Elige piedra, papel o tijera: ").lower()
    if choice_user not in ["piedra", "papel", "tijera"]:
        print("Opción no válida. Intenta de nuevo.\n")
        return None
    return choice_user

#Función para asignarle letras a la opción tomada por la CPU
def cpu_choice_to_text(choice_cpu):
    if choice_cpu == 1:
        return "piedra"
    elif choice_cpu == 2:
        return "papel"
    elif choice_cpu == 3:
        return "tijera"

#Función lógica
def comparacion(cpu_text, user_choice):
    print(f"\nCPU eligió: {cpu_text}")
    print(f"Tú elegiste: {user_choice}\n")

    if cpu_text == user_choice:
        print("¡Empate!")
        return "Empate"
    elif (user_choice == "piedra" and cpu_text == "tijera") or (user_choice == "papel" and cpu_text == "piedra") or (user_choice == "tijera" and cpu_text == "papel"):
        print("¡Ganaste!")
        return "Usuario"
    else:
        print("Perdiste...")
        return "CPU"

#Ejecución de código y llamamiento de funciones
while play_again:
    cpu_choice = juegos()
    cpu_text = cpu_choice_to_text(cpu_choice)

    user_choice = None
    while user_choice is None:
        user_choice = game()

    ganador = comparacion(cpu_text, user_choice)

    if ganador == "Usuario":
        puntos_usuario += 1
    elif ganador == "CPU":
        puntos_cpu += 1

    print(f"\nMarcador actual:")
    print(f"Tú: {puntos_usuario} puntos | CPU: {puntos_cpu} puntos\n")

    respuesta = input("¿Quieres jugar de nuevo? (si/no): ").lower()
    if respuesta != "si":
        play_again = False

#Mensaje final
print("\n¡Gracias por jugar! Resultado final:")
print(f"Tú: {puntos_usuario} puntos | CPU: {puntos_cpu} puntos")
````

### Codigo

````python
import random

#Funciones a implementar

#Función para ingresar dinero
def ask_user_bet():
    bet = float(input("\n¿Cuánto dinero desea ingresar a la máquina? $: "))
    print("Haz ingresado $", bet)
    return bet

#Función para seleccionar las palabras
def select_words():
    words = ["Cerezas", "Naranjas", "Ciruelas", "Campanas", "Melones", "Barras"]
    word1 = random.choice(words)
    word2 = random.choice(words)
    word3 = random.choice(words)
    print(word1, "|", word2, "|", word3)
    return word1, word2, word3

#Función lógica
def prize_logic(word1, word2, word3, bet):
    if word1 == word2 == word3:
        earning = bet * 3
        print("¡Felicidades, has ganado!", earning)
    elif word1 == word2 or word1 == word3 or word2 == word3:
        earning = bet * 2
        print("¡Felicidades, has ganado!", earning)
    else:
        earning = 0
        print("No ganaste nada :(")
    return earning

#Ejecución de código

#Mensajes a mostrar
print("¡Bienvenido a la máquina tragaperras!")
print("\nDeberás ingresar una cantidad de dinero que estés dispuesto a apostar")
print("- Si coinciden tres de los tres mensajes tu valor ingresado se triplicará")
print("- Si coinciden dos de los tres mensajes tu valor ingresado se duplicará")
print("- Si no coinciden ninguno de los mensajes, no ganarás nada")

total_earning = 0

#Código llamando a las funciones
while True:
    bet = ask_user_bet()
    word1, word2, word3 = select_words()
    earning = prize_logic(word1, word2, word3, bet)
    total_earning += earning

    start = input("¿Desea jugar otra vez? (si o no): ").lower()
    if start == "no":
        print("\nGracias por jugar. Ganancias totales: $", total_earning)
        break
    elif start != "si":
        print("No recibimos una respuesta correcta. Fin del juego.")
        print("\nGracias por jugar. Ganancias totales: $", total_earning)
        break
````

### Codigo

````python
# Estructuras para guardar datos
cuenta = {
    'username': None,
    'password': None,
    'money': 0,
    'historial': []
}

# Funciones
def create_account():
    username = input("Ingrese un nombre de usuario: ")
    password = input("Ingrese una contraseña: ")
    cuenta['username'] = username
    cuenta['password'] = password
    cuenta['money'] = 0
    cuenta['historial'] = []
    print("Cuenta creada exitosamente.\n")

def login_account():
    username_log = input("Ingrese su nombre de usuario: ")
    password_log = input("Ingrese su contraseña: ")
    if username_log == cuenta['username'] and password_log == cuenta['password']:
        print(f"Bienvenido, {username_log}.\n")
        return True
    else:
        print("Nombre de usuario o contraseña incorrectos.\n")
        return False

def deposit_money():
    try:
        cantidad = int(input("Ingrese una cantidad de dinero para depositar: "))
        if cantidad > 0:
            cuenta['money'] += cantidad
            cuenta['historial'].append(f"Depósito: +${cantidad}")
            print(f"Depósito exitoso. Nuevo saldo: ${cuenta['money']}\n")
        else:
            print("La cantidad debe ser mayor que 0.\n")
    except ValueError:
        print("Entrada inválida. Ingrese un número.\n")

def check_balance():
    print(f"Su saldo actual es: ${cuenta['money']}\n")

def withdraw_money():
    try:
        cantidad = int(input("Ingrese una cantidad de dinero para retirar: "))
        if cantidad <= 0:
            print("La cantidad debe ser mayor que 0.\n")
        elif cantidad > cuenta['money']:
            print("Fondos insuficientes.\n")
        else:
            cuenta['money'] -= cantidad
            cuenta['historial'].append(f"Retiro: -${cantidad}")
            print(f"Retiro exitoso. Nuevo saldo: ${cuenta['money']}\n")
    except ValueError:
        print("Entrada inválida. Ingrese un número.\n")

def show_history():
    if cuenta['historial']:
        print("Historial de transacciones:")
        for transaccion in cuenta['historial']:
            print(f"- {transaccion}")
    else:
        print("No hay transacciones registradas.\n")
    print()

# Programa principal
print("Bienvenido al Cajero Automático.")

# Lógica para crear e iniciar sesión
while True:
    if cuenta['username'] is None:
        deseo = input("No hay cuenta creada. ¿Desea crear una cuenta? (si/no): ").lower()
        if deseo == "si":
            create_account()
        else:
            print("Debe crear una cuenta para continuar.\n")
    else:
        deseo = input("¿Desea iniciar sesión? (si/no): ").lower()
        if deseo == "si":
            if login_account():
                break
        else:
            print("Debe iniciar sesión para continuar.\n")

# Menú de operaciones
while True:
    print("--- Menú de opciones ---")
    print("1. Depositar dinero")
    print("2. Consultar saldo")
    print("3. Retirar dinero")
    print("4. Ver historial de transacciones")
    print("5. Cerrar sesión")

    opcion = input("Seleccione una opción (1-5): ")
    print()

    if opcion == "1":
        deposit_money()
    elif opcion == "2":
        check_balance()
    elif opcion == "3":
        withdraw_money()
    elif opcion == "4":
        show_history()
    elif opcion == "5":
        print("Sesión cerrada. Gracias por usar el cajero automático.\n")
        break
    else:
        print("Opción no válida. Intente de nuevo.\n")
````

````python
## Codigo 4
# Sistema de gestión para la tienda Shaman Tools

# --- Datos iniciales ---
usuarios = [
    {"usuario": "Ren Tao", "contrasena": "1234_4321", "rol": "admin"},
    {"usuario": "Ioh Asakura", "contrasena": "abc_cba", "rol": "vendedor"},
    {"usuario": "Chocolove McDonnell", "contrasena": "11110000", "rol": "vendedor"}
]

inventario = [
    {"id": 1, "material": "Ladrillo 10x20x30", "precio_compra": 1200, "precio_venta": 1500, "cantidad": 1000, "unidad": "un"},
    {"id": 2, "material": "Cemento Gris x50kg", "precio_compra": 28000, "precio_venta": 30000, "cantidad": 200, "unidad": "saco"},
    {"id": 3, "material": "Varilla Corrugada #5 x6m", "precio_compra": 33000, "precio_venta": 35000, "cantidad": 150, "unidad": "un"},
    {"id": 4, "material": "Impermeabilizante 22kg", "precio_compra": 350000, "precio_venta": 360000, "cantidad": 50, "unidad": "un"},
    {"id": 5, "material": "Malla Electro Soldada 6x2.2m", "precio_compra": 42000, "precio_venta": 45000, "cantidad": 75, "unidad": "un"},
]

ventas = []

# --- Funciones comunes ---
def buscar_producto_por_id(producto_id):
    for p in inventario:
        if p["id"] == producto_id:
            return p
    return {}  # Retorna diccionario vacío si no encuentra el producto

def listar_inventario():
    for p in inventario:
        print(f'ID: {p["id"]} - {p["material"]} - Cantidad: {p["cantidad"]} {p["unidad"]}')

# --- Funciones de admin ---
def registrar_usuario(nombre, contrasena, rol):
    usuarios.append({"usuario": nombre, "contrasena": contrasena, "rol": rol})

def agregar_actualizar_producto(producto):
    existente = buscar_producto_por_id(producto["id"])
    if existente:
        existente.update(producto)
    else:
        inventario.append(producto)

def comprar_producto(producto_id, cantidad):
    producto = buscar_producto_por_id(producto_id)
    if producto:
        producto["cantidad"] += cantidad
    else:
        print("Producto no encontrado en el inventario.")

def ver_resumen_ventas():
    total = 0
    print("Resumen de ventas:")
    for venta in ventas:
        print(venta)
        total += venta["total"]
    print(f"Total generado: ${total}")

# --- Funciones de vendedor ---
def vender_producto(producto_id, cantidad):
    producto = buscar_producto_por_id(producto_id)
    if producto and producto["cantidad"] >= cantidad:
        producto["cantidad"] -= cantidad
        total = cantidad * producto["precio_venta"]
        ventas.append({
            "producto": producto["material"],
            "cantidad": cantidad,
            "total": total
        })
        print(f"Venta realizada: {cantidad} x {producto['material']} - Total: ${total}")
    else:
        print("No hay suficiente stock para esta venta.")

# --- Inicio de sesión ---
def login():
    usuario = input("Usuario: ")
    contrasena = input("Contraseña: ")
    for u in usuarios:
        if u["usuario"] == usuario and u["contrasena"] == contrasena:
            return u
    print("Credenciales incorrectas.")
    return {}

# --- Menú interactivo ---
def menu():
    user = login()
    if user:
        while True:
            print("\n--- MENÚ ---")
            if user["rol"] == "admin":
                print("1. Registrar usuario")
                print("2. Agregar/Actualizar producto")
                print("3. Consultar inventario")
                print("4. Comprar producto")
                print("5. Ver resumen de ventas")
                print("6. Salir")
                opc = input("Elige una opción: ")
                if opc == "1":
                    n = input("Nombre nuevo usuario: ")
                    c = input("Contraseña: ")
                    r = input("Rol (admin/vendedor): ")
                    registrar_usuario(n, c, r)
                elif opc == "2":
                    id_p = int(input("ID producto: "))
                    mat = input("Nombre material: ")
                    pc = int(input("Precio de compra: "))
                    pv = int(input("Precio de venta: "))
                    cant = int(input("Cantidad: "))
                    uni = input("Unidad: ")
                    agregar_actualizar_producto({"id": id_p, "material": mat, "precio_compra": pc, "precio_venta": pv, "cantidad": cant, "unidad": uni})
                elif opc == "3":
                    listar_inventario()
                elif opc == "4":
                    idp = int(input("ID del producto a comprar: "))
                    cant = int(input("Cantidad a agregar: "))
                    comprar_producto(idp, cant)
                elif opc == "5":
                    ver_resumen_ventas()
                elif opc == "6":
                    break
            elif user["rol"] == "vendedor":
                print("1. Consultar inventario")
                print("2. Vender producto")
                print("3. Salir")
                opc = input("Elige una opción: ")
                if opc == "1":
                    listar_inventario()
                elif opc == "2":
                    idp = int(input("ID producto a vender: "))
                    cant = int(input("Cantidad a vender: "))
                    vender_producto(idp, cant)
                elif opc == "3":
                    break

# Ejecutar menú
menu()
````
